## Updated implementation architecture

The Assessment 2 pipeline below updates the proposal architecture to reflect implementation evidence:
**raw files → validation → cleaning/unit harmonisation → daily aggregation → spatial fire engineering → station-day merge → leakage check → three models → validation/test → explainability → saved outputs**.

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Report-ready architecture figure generated from the implemented pipeline.
stages = [
    "NT EPA hourly\n+ FIRMS S-NPP",
    "Raw validation\ncoverage / schema",
    "Cleaning\nunit harmonisation",
    "Daily + spatial\nfeature engineering",
    "Station-day\nmerged table",
    "Leakage-safe\nt → t+1 targets",
    "3 model tasks\nClass / Reg / Count",
    "2023 validation\n2024 final test",
    "SHAP + figures\n+ saved evidence",
]

fig, ax = plt.subplots(figsize=(18, 3.3))
ax.axis("off")
xs = np.linspace(0.04, 0.96, len(stages))
for i, (x, label) in enumerate(zip(xs, stages)):
    ax.text(
        x, 0.5, label, ha="center", va="center",
        bbox=dict(boxstyle="round,pad=0.45", fc="white", ec="black"),
        transform=ax.transAxes, fontsize=9
    )
    if i < len(stages) - 1:
        ax.annotate(
            "", xy=(xs[i+1] - 0.045, 0.5), xytext=(x + 0.045, 0.5),
            xycoords=ax.transAxes, textcoords=ax.transAxes,
            arrowprops=dict(arrowstyle="->")
        )
ax.set_title("Fire2Air Darwin — Updated Assessment 2 Architecture", pad=14)
fig.tight_layout()
fig.savefig(FIGURE_DIR / "01_updated_architecture.png", dpi=180, bbox_inches="tight")
plt.show()
```
